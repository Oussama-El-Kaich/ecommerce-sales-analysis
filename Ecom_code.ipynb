### Step 1: Loading the Dataset
The dataset used in this analysis is the **Online Retail II** dataset from the **UCI Machine Learning Repository**.  
It contains transactional data for a UK-based online gift retailer from 2010 to 2011.


```python
# --- Step 1: Load Dataset ---
# Read the 2010–2011 sheet from the Online Retail II dataset
retail = pd.read_excel('/content/drive/My Drive/Final/online_retail_zwei.xlsx')
```

### Step 2: Creating a Working Copy
A copy of the dataset is created to ensure that the original data remains unchanged during cleaning and transformation.


```python
# --- Step 2: Create Working Copy ---
df = retail.copy()
```

### Step 3: Data Cleaning
This section ensures that the dataset contains only **valid transactions**:
- Converts invoice numbers to string format.  
- Removes cancelled orders (invoices beginning with “C”).  
- Filters out negative or zero quantities and prices.  
- Drops records missing essential values such as *Customer ID*, *Description*, or *InvoiceDate*.


```python
# --- Step 3: Data Cleaning ---
df['Invoice'] = df['Invoice'].astype(str)
df = df[~df['Invoice'].str.startswith('C')]
df = df[(df['Quantity'] > 0) & (df['Price'] > 0)]
df = df.dropna(subset=['Description', 'Customer ID', 'InvoiceDate'])
```

### Step 4: Renaming Columns
Columns are renamed for clarity and consistency.  
For example, *StockCode → ProductCode* and *Price → UnitPrice*.


```python
# --- Step 4: Rename Columns for Clarity ---
df.rename(columns={
    'StockCode': 'ProductCode',
    'Description': 'ProductName',
    'InvoiceDate': 'Date',
    'Price': 'UnitPrice',
    'Customer ID': 'CustomerID'
}, inplace=True)

print("Columns renamed successfully:")
print(df.columns)
```

### Step 5: Feature Engineering
Additional variables are created to enrich analysis:
- **TotalSales:** computed as *Quantity × UnitPrice*  
- **Month** and **Weekday:** extracted from the invoice date to enable temporal and behavioral analysis.


```python
# --- Step 5: Feature Engineering ---
df['TotalSales'] = df['Quantity'] * df['UnitPrice']
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Month'] = df['Date'].dt.to_period('M').astype(str)
df['Weekday'] = df['Date'].dt.day_name()
```

### Step 6: Exporting Cleaned Data
The final cleaned dataset is saved to Google Drive as **online_retail_clean.csv**.  
This file is used later for visualization and dashboard creation in **Qlik Sense**.


```python
# --- Step 6: Save Clean Dataset for Qlik Analysis ---
df.to_csv('/content/drive/My Drive/Final/online_retail_zwei_clean.csv', index=False)

print("✅ Cleaning complete! File saved to: /content/drive/My Drive/Final/online_retail_zwei_clean.csv")
print("Remaining rows:", len(df))
print("Unique customers:", df['CustomerID'].nunique())

df.head()
```
